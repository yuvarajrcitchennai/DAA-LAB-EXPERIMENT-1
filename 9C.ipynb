```python
import random

CAPACITY = 1.0
NUM_ITEMS = 50
TRIALS = 100


# Next Fit Decreasing (NFD)
def next_fit_decreasing(items, capacity=1.0):
    items = sorted(items, reverse=True)

    bins = 1
    remaining = capacity

    for item in items:
        if item <= remaining:
            remaining -= item
        else:
            bins += 1
            remaining = capacity - item

    return bins


# First Fit Decreasing (FFD)
def first_fit_decreasing(items, capacity=1.0):
    items = sorted(items, reverse=True)

    bins = []

    for item in items:
        placed = False

        for i in range(len(bins)):
            if bins[i] >= item:
                bins[i] -= item
                placed = True
                break

        if not placed:
            bins.append(capacity - item)

    return len(bins)


# Best Fit Decreasing (BFD)
def best_fit_decreasing(items, capacity=1.0):
    items = sorted(items, reverse=True)

    bins = []

    for item in items:
        best_idx = -1
        best_space = float("inf")

        for i in range(len(bins)):
            if bins[i] >= item and (bins[i] - item) < best_space:
                best_space = bins[i] - item
                best_idx = i

        if best_idx != -1:
            bins[best_idx] -= item
        else:
            bins.append(capacity - item)

    return len(bins)


total_nfd = 0
total_ffd = 0
total_bfd = 0

for _ in range(TRIALS):

    items = [
        round(random.uniform(0.1, 0.9), 2)
        for _ in range(NUM_ITEMS)
    ]

    total_nfd += next_fit_decreasing(items, CAPACITY)
    total_ffd += first_fit_decreasing(items, CAPACITY)
    total_bfd += best_fit_decreasing(items, CAPACITY)

avg_nfd = total_nfd / TRIALS
avg_ffd = total_ffd / TRIALS
avg_bfd = total_bfd / TRIALS

print("After", TRIALS, "Trials")
print("Average Bins Used")
print("-------------------")
print("NFD =", round(avg_nfd, 2))
print("FFD =", round(avg_ffd, 2))
print("BFD =", round(avg_bfd, 2))
```
