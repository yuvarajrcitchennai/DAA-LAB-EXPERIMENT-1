```python
import heapq

# Modified Dijkstra's Algorithm
def dijkstra(graph, source):
    n = len(graph)

    dist = [float('inf')] * n
    hops = [float('inf')] * n
    parent = [-1] * n

    dist[source] = 0
    hops[source] = 0

    # (distance, hops, vertex)
    pq = [(0, 0, source)]

    while pq:
        d, h, u = heapq.heappop(pq)

        if d > dist[u]:
            continue

        for v, w in graph[u]:
            new_dist = dist[u] + w
            new_hops = hops[u] + 1

            # Update if shorter distance OR same distance with fewer hops
            if (new_dist < dist[v]) or \
               (new_dist == dist[v] and new_hops < hops[v]):

                dist[v] = new_dist
                hops[v] = new_hops
                parent[v] = u

                heapq.heappush(pq, (new_dist, new_hops, v))

    return dist, hops, parent


# Reconstruct Path
def get_path(parent, target):
    path = []

    while target != -1:
        path.append(target)
        target = parent[target]

    return path[::-1]


# Graph with 6 vertices
graph = {
    0: [(1, 2), (2, 4)],
    1: [(3, 2), (4, 5)],
    2: [(3, 2)],
    3: [(5, 3)],
    4: [(5, 2)],
    5: []
}

source = 0

dist, hops, parent = dijkstra(graph, source)

print("Vertex\tDistance\tHops\tPath")
print("-" * 40)

for v in range(len(graph)):
    path = get_path(parent, v)
    print(f"{v}\t{dist[v]}\t\t{hops[v]}\t{' -> '.join(map(str, path))}")
```
