```python
# Cloud Server Resource Allocation using
# First Fit Decreasing (FFD) and Best Fit Decreasing (BFD)

def first_fit_decreasing(tasks, capacity):
    tasks = sorted(tasks, reverse=True)

    servers = []
    remaining = []

    for task in tasks:
        placed = False

        for i in range(len(remaining)):
            if remaining[i] >= task:
                remaining[i] -= task
                servers[i].append(task)
                placed = True
                break

        if not placed:
            remaining.append(capacity - task)
            servers.append([task])

    return servers


def best_fit_decreasing(tasks, capacity):
    tasks = sorted(tasks, reverse=True)

    servers = []
    remaining = []

    for task in tasks:
        best_idx = -1
        best_space = float('inf')

        for i in range(len(remaining)):
            if remaining[i] >= task and remaining[i] - task < best_space:
                best_space = remaining[i] - task
                best_idx = i

        if best_idx != -1:
            remaining[best_idx] -= task
            servers[best_idx].append(task)
        else:
            remaining.append(capacity - task)
            servers.append([task])

    return servers


tasks = [1.5, 2.0, 3.5, 0.5, 4.0,
         1.0, 2.5, 3.0, 1.5, 2.0,
         0.5, 4.0, 1.0, 3.5, 2.0]

capacity = 8.0

ffd_servers = first_fit_decreasing(tasks, capacity)
bfd_servers = best_fit_decreasing(tasks, capacity)

print("FFD Allocation")
for i, s in enumerate(ffd_servers, 1):
    print(f"Server {i}: {s}  Used={sum(s)} GB")

print("\nBFD Allocation")
for i, s in enumerate(bfd_servers, 1):
    print(f"Server {i}: {s}  Used={sum(s)} GB")

print("\nFFD Servers Needed =", len(ffd_servers))
print("BFD Servers Needed =", len(bfd_servers))
```
