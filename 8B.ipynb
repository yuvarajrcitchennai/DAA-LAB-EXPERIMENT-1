```python
from itertools import permutations

INF = float('inf')

# 6-City Distance Matrix (Road Distances in km)
cost = [
    [INF, 10, 15, 20, 25, 18],  # A
    [10, INF, 35, 25, 17, 28],  # B
    [15, 35, INF, 30, 28, 40],  # C
    [20, 25, 30, INF, 22, 12],  # D
    [25, 17, 28, 22, INF, 16],  # E
    [18, 28, 40, 12, 16, INF]   # F
]

cities = ['A', 'B', 'C', 'D', 'E', 'F']
n = len(cost)


# Brute Force TSP
def tsp_brute_force(cost):
    best_cost = INF
    best_path = None

    for perm in permutations(range(1, n)):
        path = [0] + list(perm) + [0]

        total = 0
        for i in range(n):
            total += cost[path[i]][path[i + 1]]

        if total < best_cost:
            best_cost = total
            best_path = path

    return best_path, best_cost


# Nearest Neighbour Heuristic
def nearest_neighbour(cost):
    visited = [False] * n
    path = [0]
    visited[0] = True
    total_cost = 0
    current = 0

    for _ in range(n - 1):
        nearest = -1
        min_cost = INF

        for city in range(n):
            if not visited[city] and cost[current][city] < min_cost:
                min_cost = cost[current][city]
                nearest = city

        path.append(nearest)
        visited[nearest] = True
        total_cost += min_cost
        current = nearest

    path.append(0)
    total_cost += cost[current][0]

    return path, total_cost


# Run both methods
bf_path, bf_cost = tsp_brute_force(cost)
nn_path, nn_cost = nearest_neighbour(cost)

print("Brute Force Optimal Route:")
print(" -> ".join(cities[i] for i in bf_path))
print("Minimum Cost =", bf_cost)

print("\nNearest Neighbour Route:")
print(" -> ".join(cities[i] for i in nn_path))
print("Heuristic Cost =", nn_cost)
```
