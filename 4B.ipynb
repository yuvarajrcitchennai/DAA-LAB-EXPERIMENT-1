```python
import heapq

# Dijkstra's Algorithm
def dijkstra(graph, source):
    n = len(graph)
    dist = [float('inf')] * n
    parent = [-1] * n

    dist[source] = 0
    pq = [(0, source)]  # (distance, node)

    while pq:
        current_dist, u = heapq.heappop(pq)

        if current_dist > dist[u]:
            continue

        for v, weight in graph[u]:
            if dist[u] + weight < dist[v]:
                dist[v] = dist[u] + weight
                parent[v] = u
                heapq.heappush(pq, (dist[v], v))

    return dist, parent


# Reconstruct shortest path
def get_path(parent, target):
    path = []

    while target != -1:
        path.append(target)
        target = parent[target]

    return path[::-1]


# Directed Graph (One-way roads)
graph = {
    0: [(1, 10), (2, 3)],      # Airport
    1: [(2, 1), (3, 2)],
    2: [(1, 4), (3, 8), (4, 2)],
    3: [(4, 7)],
    4: []                      # Central Station
}

source = 0
destination = 4

dist, parent = dijkstra(graph, source)
path = get_path(parent, destination)

print("Fastest Route from Airport (0) to Central Station (4):")
print(" -> ".join(map(str, path)))
print("Total Travel Time:", dist[destination], "minutes")
```
