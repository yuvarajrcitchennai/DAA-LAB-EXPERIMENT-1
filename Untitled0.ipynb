```python
# N-Queens using Bit Manipulation (Bitmask)
# O(1) safety check for columns and diagonals

def solve_n_queens_bitmask(n):
    solutions = 0
    backtracks = 0

    def backtrack(cols, diag1, diag2):
        nonlocal solutions, backtracks

        if cols == (1 << n) - 1:
            solutions += 1
            return

        available = ((1 << n) - 1) & ~(cols | diag1 | diag2)

        while available:
            pos = available & -available
            available -= pos

            backtracks += 1

            backtrack(
                cols | pos,
                (diag1 | pos) << 1,
                (diag2 | pos) >> 1
            )

    backtrack(0, 0, 0)
    return solutions, backtracks


# Test for N = 12
n = 12

solutions, backtracks = solve_n_queens_bitmask(n)

print("N =", n)
print("Number of Solutions =", solutions)
print("Total Backtracks =", backtracks)
```
