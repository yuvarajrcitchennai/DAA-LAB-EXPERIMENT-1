```python
import random
import heapq
import time

# Generate Dense Graph (100 vertices)
V = 100
INF = float('inf')

# Adjacency Matrix
matrix = [[0] * V for _ in range(V)]

for i in range(V):
    for j in range(i + 1, V):
        weight = random.randint(1, 20)
        matrix[i][j] = weight
        matrix[j][i] = weight

# Convert Matrix to Adjacency List
adj = {i: [] for i in range(V)}
for i in range(V):
    for j in range(V):
        if matrix[i][j] != 0:
            adj[i].append((j, matrix[i][j]))


# Prim's Algorithm using Adjacency Matrix
def prim_matrix(graph):
    V = len(graph)
    selected = [False] * V
    key = [INF] * V
    parent = [-1] * V

    key[0] = 0
    cost = 0

    for _ in range(V):
        u = -1
        minimum = INF

        for v in range(V):
            if not selected[v] and key[v] < minimum:
                minimum = key[v]
                u = v

        selected[u] = True
        cost += key[u]

        for v in range(V):
            if graph[u][v] != 0 and not selected[v] and graph[u][v] < key[v]:
                key[v] = graph[u][v]
                parent[v] = u

    return cost


# Prim's Algorithm using Heap (Adjacency List)
def prim_heap(adj):
    V = len(adj)
    visited = [False] * V
    pq = [(0, 0)]
    cost = 0

    while pq:
        weight, u = heapq.heappop(pq)

        if visited[u]:
            continue

        visited[u] = True
        cost += weight

        for v, w in adj[u]:
            if not visited[v]:
                heapq.heappush(pq, (w, v))

    return cost


# Compare Execution Time
start = time.perf_counter()
cost1 = prim_matrix(matrix)
time_matrix = time.perf_counter() - start

start = time.perf_counter()
cost2 = prim_heap(adj)
time_heap = time.perf_counter() - start

print("Prim's using Adjacency Matrix")
print("MST Cost:", cost1)
print("Execution Time:", round(time_matrix, 6), "seconds")

print("\nPrim's using Heap (Adjacency List)")
print("MST Cost:", cost2)
print("Execution Time:", round(time_heap, 6), "seconds")
```
