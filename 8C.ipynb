```python
import random
import math
from itertools import permutations

# -----------------------------
# Generate 10 random cities
# -----------------------------
random.seed(42)

N = 10

cities = [
    (random.randint(0, 100), random.randint(0, 100))
    for _ in range(N)
]

# Distance matrix
dist = [[0] * N for _ in range(N)]

for i in range(N):
    for j in range(N):
        if i != j:
            dist[i][j] = round(
                math.sqrt(
                    (cities[i][0] - cities[j][0]) ** 2 +
                    (cities[i][1] - cities[j][1]) ** 2
                )
            )


# -----------------------------
# Tour Cost
# -----------------------------
def tour_cost(tour):
    cost = 0

    for i in range(len(tour) - 1):
        cost += dist[tour[i]][tour[i + 1]]

    return cost


# -----------------------------
# Nearest Neighbour
# -----------------------------
def nearest_neighbour():
    visited = [False] * N

    tour = [0]
    visited[0] = True
    current = 0

    while len(tour) < N:
        nearest = None
        best = float('inf')

        for city in range(N):
            if not visited[city] and dist[current][city] < best:
                best = dist[current][city]
                nearest = city

        tour.append(nearest)
        visited[nearest] = True
        current = nearest

    tour.append(0)
    return tour


# -----------------------------
# 2-OPT Improvement
# -----------------------------
def two_opt(tour):
    improved = True

    while improved:
        improved = False

        for i in range(1, len(tour) - 2):
            for j in range(i + 1, len(tour) - 1):

                if j - i == 1:
                    continue

                new_tour = (
                    tour[:i]
                    + tour[i:j][::-1]
                    + tour[j:]
                )

                if tour_cost(new_tour) < tour_cost(tour):
                    tour = new_tour
                    improved = True

    return tour


# -----------------------------
# Exact Optimal (Brute Force)
# Warning: works for small N.
# For N=10 it may take time.
# -----------------------------
def brute_force_tsp():
    best_cost = float('inf')
    best_tour = None

    for perm in permutations(range(1, N)):
        tour = [0] + list(perm) + [0]

        c = tour_cost(tour)

        if c < best_cost:
            best_cost = c
            best_tour = tour

    return best_tour, best_cost


# -----------------------------
# Run Algorithms
# -----------------------------
nn_tour = nearest_neighbour()
nn_cost = tour_cost(nn_tour)

opt_tour = two_opt(nn_tour.copy())
opt_cost = tour_cost(opt_tour)

print("Nearest Neighbour Tour:")
print(nn_tour)
print("Cost =", nn_cost)

print("\n2-OPT Improved Tour:")
print(opt_tour)
print("Cost =", opt_cost)

improvement = ((nn_cost - opt_cost) / nn_cost) * 100

print("\nImprovement =", round(improvement, 2), "%")
```
