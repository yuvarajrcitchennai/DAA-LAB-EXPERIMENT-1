```python
import random

# Interpolation Search for Floating-Point Numbers
def interpolation_search(arr, key):
    low = 0
    high = len(arr) - 1
    comparisons = 0

    while low <= high and arr[low] <= key <= arr[high]:
        comparisons += 1

        if arr[high] == arr[low]:
            if arr[low] == key:
                return low, comparisons
            else:
                return -1, comparisons

        # Estimate the position
        pos = low + int(((key - arr[low]) * (high - low)) /
                        (arr[high] - arr[low]))

        if pos < low or pos > high:
            break

        comparisons += 1
        if arr[pos] == key:
            return pos, comparisons

        comparisons += 1
        if arr[pos] < key:
            low = pos + 1
        else:
            high = pos - 1

    return -1, comparisons


# Dataset sizes
sizes = [10000, 50000, 100000]

print("Interpolation Search on Floating-Point Numbers")
print("-" * 60)

for n in sizes:
    # Generate uniformly distributed floating-point numbers
    arr = sorted(random.uniform(0.0, 1000.0) for _ in range(n))

    # Choose a key that exists in the array
    key = arr[random.randint(0, n - 1)]

    index, comparisons = interpolation_search(arr, key)

    print(f"Dataset Size : {n}")
    print(f"Key Searched : {key:.4f}")
    print(f"Found at Index : {index}")
    print(f"Comparisons : {comparisons}")
    print("-" * 60)
```
