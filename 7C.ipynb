```python
# Knight's Tour using Backtracking
# Start Position : (0,0)
# Board Size : 8 x 8

N = 8

# Possible knight moves
move_x = [2, 1, -1, -2, -2, -1, 1, 2]
move_y = [1, 2, 2, 1, -1, -2, -2, -1]


def is_safe(x, y, board):
    return (0 <= x < N and
            0 <= y < N and
            board[x][y] == -1)


def solve_knights_tour():
    board = [[-1 for _ in range(N)] for _ in range(N)]

    # Starting position
    board[0][0] = 0

    if not solve_util(0, 0, 1, board):
        print("No solution exists")
        return None

    return board


def solve_util(x, y, move_no, board):

    # All 64 cells visited
    if move_no == N * N:
        return True

    for k in range(8):
        next_x = x + move_x[k]
        next_y = y + move_y[k]

        if is_safe(next_x, next_y, board):
            board[next_x][next_y] = move_no

            if solve_util(next_x, next_y, move_no + 1, board):
                return True

            # Backtrack
            board[next_x][next_y] = -1

    return False


def print_board(board):
    print("\nKnight's Tour Solution:\n")

    for row in board:
        for cell in row:
            print(f"{cell:2}", end=" ")
        print()


solution = solve_knights_tour()

if solution:
    print_board(solution)
```
