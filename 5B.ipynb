```python
comparison_count = 0

def min_max_second(arr):
    global comparison_count

    # Initialize
    min1 = min2 = float('inf')
    max1 = max2 = float('-inf')

    for num in arr:
        # Minimum and Second Minimum
        comparison_count += 1
        if num < min1:
            min2 = min1
            min1 = num
        else:
            comparison_count += 1
            if num < min2:
                min2 = num

        # Maximum and Second Maximum
        comparison_count += 1
        if num > max1:
            max2 = max1
            max1 = num
        else:
            comparison_count += 1
            if num > max2:
                max2 = num

    return min1, min2, max1, max2


# Main Program
arr = [12, 5, 8, 1, 19, 3, 25, 7, 15, 2]

comparison_count = 0

minimum, second_minimum, maximum, second_maximum = min_max_second(arr)

print("Array:", arr)
print("Minimum:", minimum)
print("Second Minimum:", second_minimum)
print("Maximum:", maximum)
print("Second Maximum:", second_maximum)
print("Total Comparisons:", comparison_count)
```
