```python
import random

# Generate a random 10x10 text matrix
rows, cols = 10, 10
text = [[random.randint(0, 9) for _ in range(cols)] for _ in range(rows)]

# Define a 3x3 pattern matrix
pattern = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# (Optional) Insert the pattern into the text matrix for testing
start_row, start_col = 4, 5
for i in range(3):
    for j in range(3):
        text[start_row + i][start_col + j] = pattern[i][j]

# Naive 2D Pattern Matching
comparisons = 0
matches = []

for i in range(rows - 3 + 1):
    for j in range(cols - 3 + 1):
        found = True

        for x in range(3):
            for y in range(3):
                comparisons += 1

                if text[i + x][j + y] != pattern[x][y]:
                    found = False
                    break

            if not found:
                break

        if found:
            matches.append((i, j))

# Display Text Matrix
print("Text Matrix:")
for row in text:
    print(row)

# Display Pattern Matrix
print("\nPattern Matrix:")
for row in pattern:
    print(row)

# Display Results
print("\nPattern Found At:", matches)
print("Total Element Comparisons:", comparisons)
```
