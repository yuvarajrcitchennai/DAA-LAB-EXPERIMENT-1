```python
# Kruskal's Algorithm using Union-Find

class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, x, y):
        rx = self.find(x)
        ry = self.find(y)

        if rx == ry:
            return False

        if self.rank[rx] < self.rank[ry]:
            self.parent[rx] = ry
        elif self.rank[rx] > self.rank[ry]:
            self.parent[ry] = rx
        else:
            self.parent[ry] = rx
            self.rank[rx] += 1

        return True


def kruskal(vertices, edges):
    edges.sort(key=lambda x: x[2])

    uf = UnionFind(len(vertices))
    vertex_index = {v: i for i, v in enumerate(vertices)}

    mst = []
    total_cost = 0

    for u, v, w in edges:
        if uf.union(vertex_index[u], vertex_index[v]):
            mst.append((u, v, w))
            total_cost += w

    return mst, total_cost


# Vertices
vertices = ['A', 'B', 'C', 'D', 'E', 'F']

# Edges (City1, City2, Cost)
edges = [
    ('A', 'B', 4),
    ('A', 'C', 2),
    ('B', 'C', 5),
    ('B', 'D', 10),
    ('C', 'E', 3),
    ('D', 'E', 7),
    ('D', 'F', 8),
    ('E', 'F', 6)
]

mst, cost = kruskal(vertices, edges)

print("Minimum Spanning Tree (Kruskal):")
for u, v, w in mst:
    print(f"{u} -- {v} : {w}")

print("\nTotal Cost =", cost)
```
