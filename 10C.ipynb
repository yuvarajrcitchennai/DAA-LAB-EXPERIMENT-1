```python
import random
import math
import time
import heapq
import sys

sys.setrecursionlimit(30000)

# ----------------------------------
# Standard Quick Sort
# ----------------------------------
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


def quick_sort(arr, low, high):
    if low < high:
        pi = partition(arr, low, high)

        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)


# ----------------------------------
# Heap Sort
# ----------------------------------
def heap_sort_subarray(arr, start, end):
    temp = arr[start:end + 1]

    heapq.heapify(temp)

    for i in range(start, end + 1):
        arr[i] = heapq.heappop(temp)


# ----------------------------------
# Insertion Sort
# ----------------------------------
def insertion_sort(arr, low, high):
    for i in range(low + 1, high + 1):
        key = arr[i]
        j = i - 1

        while j >= low and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key


# ----------------------------------
# IntroSort Utility
# ----------------------------------
def introsort_util(arr, low, high, depth_limit):

    size = high - low + 1

    # Small subarray → Insertion Sort
    if size < 16:
        insertion_sort(arr, low, high)
        return

    # Depth exceeded → Heap Sort
    if depth_limit == 0:
        heap_sort_subarray(arr, low, high)
        return

    if low < high:
        pi = partition(arr, low, high)

        introsort_util(arr, low, pi - 1, depth_limit - 1)
        introsort_util(arr, pi + 1, high, depth_limit - 1)


# ----------------------------------
# IntroSort
# ----------------------------------
def introsort(arr):
    n = len(arr)

    depth_limit = 2 * int(math.log2(n))

    introsort_util(arr, 0, n - 1, depth_limit)


# ----------------------------------
# Timing Function
# ----------------------------------
def run_test(name, sort_func, arr):
    a = arr[:]

    start = time.perf_counter()

    if name == "QuickSort":
        sort_func(a, 0, len(a) - 1)
    else:
        sort_func(a)

    elapsed = (time.perf_counter() - start) * 1000

    return elapsed


# ----------------------------------
# Test Arrays
# ----------------------------------
N = 10000

random_arr = [random.randint(1, 100000) for _ in range(N)]
sorted_arr = list(range(N))

print("-" * 60)
print(f'{"Input":<15} {"Quick Sort(ms)":>18} {"IntroSort(ms)":>18}')
print("-" * 60)

for name, arr in {
    "Random": random_arr,
    "Sorted": sorted_arr
}.items():

    try:
        quick_time = run_test(
            "QuickSort",
            quick_sort,
            arr
        )
    except RecursionError:
        quick_time = float("inf")

    intro_time = run_test(
        "IntroSort",
        introsort,
        arr
    )

    print(
        f'{name:<15}'
        f'{quick_time:>18.2f}'
        f'{intro_time:>18.2f}'
    )
```
