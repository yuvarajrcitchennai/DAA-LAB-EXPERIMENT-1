```python
# Matrix Chain Multiplication
# Top-Down (Memoization) vs Bottom-Up (DP)

def memoized_mcm(p):
    n = len(p) - 1

    memo = [[None] * (n + 1) for _ in range(n + 1)]

    def solve(i, j):
        if i == j:
            return 0

        if memo[i][j] is not None:
            return memo[i][j]

        memo[i][j] = float('inf')

        for k in range(i, j):
            cost = (
                solve(i, k)
                + solve(k + 1, j)
                + p[i - 1] * p[k] * p[j]
            )

            memo[i][j] = min(memo[i][j], cost)

        return memo[i][j]

    return solve(1, n)


def bottom_up_mcm(p):
    n = len(p) - 1

    dp = [[0] * (n + 1) for _ in range(n + 1)]

    for length in range(2, n + 1):
        for i in range(1, n - length + 2):
            j = i + length - 1

            dp[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + p[i - 1] * p[k] * p[j]
                )

                dp[i][j] = min(dp[i][j], cost)

    return dp[1][n]


# Chain of 6 matrices
# A1=30x35, A2=35x15, A3=15x5,
# A4=5x10, A5=10x20, A6=20x25

dims = [30, 35, 15, 5, 10, 20, 25]

top_down_cost = memoized_mcm(dims)
bottom_up_cost = bottom_up_mcm(dims)

print("Top-Down (Memoization) Cost :", top_down_cost)
print("Bottom-Up (DP) Cost         :", bottom_up_cost)

if top_down_cost == bottom_up_cost:
    print("Result: Both methods give the SAME minimum cost.")
else:
    print("Result: Costs are different.")
```
